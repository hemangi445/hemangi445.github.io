```python
#Load Libraries and Business Dataset

import pandas as pd
import altair as alt

alt.data_transformers.disable_max_rows()

df = pd.read_csv("iowa_retail_business_registrations_sample.csv")

df.head()
```

```python
#Prepare Business Data for Dashboard

df["Date Permit was Issued"] = pd.to_datetime(df["Date Permit was Issued"], errors="coerce")
df["Issue Year"] = df["Date Permit was Issued"].dt.year

df_dashboard = df.dropna(subset=[
    "Active or Inactive Business",
    "County Name",
    "Issue Year"
])

df_dashboard = df_dashboard[df_dashboard["Issue Year"] >= 2000]
df_dashboard = df_dashboard[df_dashboard["Issue Year"] <= 2025]

df_dashboard.head()
```

```python
selection = alt.selection_point(fields=["Active or Inactive Business"])

chart1 = alt.Chart(df_dashboard).mark_bar().encode(
    x=alt.X("Active or Inactive Business:N", title="Business Status"),
    y=alt.Y("count():Q", title="Number of Businesses"),
    color=alt.Color("Active or Inactive Business:N", title="Business Status"),
    opacity=alt.condition(selection, alt.value(1), alt.value(0.25)),
    tooltip=[
        alt.Tooltip("Active or Inactive Business:N", title="Business Status"),
        alt.Tooltip("count():Q", title="Number of Businesses")
    ]
).add_params(
    selection
).properties(
    title="Active vs Inactive Businesses",
    width=260,
    height=320
)

chart2 = alt.Chart(df_dashboard).mark_line(point=True).encode(
    x=alt.X("Issue Year:O", sort="ascending", title="Year"),
    y=alt.Y("count():Q", title="Number of Businesses"),
    color=alt.Color("Active or Inactive Business:N", title="Business Status"),
    tooltip=[
        alt.Tooltip("Issue Year:O", title="Year"),
        alt.Tooltip("Active or Inactive Business:N", title="Business Status"),
        alt.Tooltip("count():Q", title="Number of Businesses")
    ]
).transform_filter(
    selection
).properties(
    title="Business Registrations Over Time",
    width=650,
    height=320
)

dashboard = chart1 | chart2

dashboard
```

```python
dashboard.save("../assets/json/final_dashboard.json")
```

```python
#Load and Clean Census Population Data

df_pop = pd.read_csv("ACSDT5Y2023.B01003-Data.csv", skiprows=1)

# Rename all 5 columns
df_pop.columns = ["GEO_ID", "County", "Population", "Margin", "Extra"]

# Remove the first row that contains labels like "Geography"
df_pop = df_pop.iloc[1:]

# Keep only the useful columns
df_pop = df_pop[["County", "Population"]]

# Clean county names
df_pop["County"] = df_pop["County"].str.replace(" County, Iowa", "", regex=False)

# Convert population to number
df_pop["Population"] = pd.to_numeric(df_pop["Population"], errors="coerce")

# Drop missing rows
df_pop = df_pop.dropna()

df_pop.head()
```

```python
# Rename column to match
df_pop = df_pop.rename(columns={"County": "County Name"})

#Merge Business and Population Data
#I rename the county column so both datasets use the same field name, then merge the population data with the business registration data.

df_merged = df_dashboard.merge(df_pop, on="County Name", how="left")

df_merged.head()
```

```python
# Group data by County
#I group the merged data by county to calculate the number of business records in each county and keep the population value for context.

df_county = df_merged.groupby("County Name").agg({
    "Population": "first",
    "Active or Inactive Business": "count"
}).reset_index()

df_county = df_county.rename(columns={
    "Active or Inactive Business": "Business Count"
})

df_county.head()
```

```python
#top counties by business registrations

df_county_top = df_county.sort_values(

    by="Business Count",

    ascending=False

).head(15)

chart3 = alt.Chart(df_county_top).mark_bar().encode(

    x=alt.X("Business Count:Q", title="Number of Businesses"),

    y=alt.Y("County Name:N", sort="-x", title="County"),

    color=alt.Color(

        "Population:Q",

        title="Population",

        scale=alt.Scale(scheme="blues")

    ),

    tooltip=["County Name", "Business Count", "Population"]

).properties(

    title="Top 15 Counties by Business Registrations",

    width=600,

    height=400

)

chart3
```

```python
# Context chart 2: top counties by population

df_population_top = df_pop.sort_values(by="Population", ascending=False).head(15)

chart4 = alt.Chart(df_population_top).mark_bar().encode(
    x=alt.X("Population:Q", title="Population"),
    y=alt.Y("County Name:N", sort="-x", title="County"),
    tooltip=["County Name", "Population"]
).properties(
    title="Top 15 Iowa Counties by Population",
    width=600,
    height=400
)

chart4
```

```python
# Save all visualization JSON files
dashboard.save("../assets/json/final_dashboard.json")
chart3.save("../assets/json/context_business_count.json")
chart4.save("../assets/json/context_population.json")

# Save cleaned population dataset with better name
df_pop.to_csv("iowa_county_population.csv", index=False)
```
